## **Pruebas Formulas Según el CST**

```python
# Imports
from datetime import datetime
```

```python
salario_base: int = 1750000 # $COP
quincena: float = salario_base / 2
auxilio_transporte: int = 250000 # $ COP
base_liquidada: int = salario_base + auxilio_transporte
dias_laborados: int = 180 # Dias

cesantias: float = (base_liquidada * dias_laborados) / 360
intereses_cesantias: float = (cesantias * dias_laborados * 0.12) / 360
prima: float = (base_liquidada * dias_laborados) / 360
vacaciones: float = (quincena * dias_laborados) / 360

total_liquidacion: float = cesantias + vacaciones + intereses_cesantias + prima

print(f"Total a pagar: {total_liquidacion}")
```
